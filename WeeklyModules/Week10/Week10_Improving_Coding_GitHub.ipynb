# Week 10 Improving Coding — Week 6 Participation

This version keeps the original Week 6 task and uses the **MSSP6070 GitHub repository** for the CSV data. The changes improve readability and add simple checks.


## 1. Connect to the GitHub repository

This code clones the repository without putting a personal token in the notebook. It expects the repository to be public or otherwise accessible. The clone is a temporary working copy in Colab; changes are **not automatically pushed** to GitHub.


```python
from pathlib import Path
import subprocess
import pandas as pd

owner = 'shuob3344'
repository = 'MSSP6070'
repo_url = f'https://github.com/{owner}/{repository}.git'
repo_path = Path('/content') / repository

if not repo_path.exists():
    subprocess.run(['git', 'clone', repo_url, str(repo_path)], check=True)
else:
    print('Using existing repository folder:', repo_path)

data_path = repo_path / 'data' / 'Student Performance Data.csv'
if not data_path.exists():
    raise FileNotFoundError(f'Cannot find the CSV in GitHub repository: {data_path}')

df = pd.read_csv(data_path)
print('Original data shape:', df.shape)
df.head()
```

## 2. Add a math level column

As in Week 6, math scores of 80 or above are labeled `High`; all other scores are `Other`.


```python
def add_math_level(data):
    if 'math score' not in data.columns:
        raise KeyError("Missing column: math score")

    result = data.copy()
    result['math level'] = result['math score'].apply(
        lambda score: 'High' if score >= 80 else 'Other'
    )
    return result


new_df = add_math_level(df)
new_df.head()
```

## 3. Check the updated data


```python
print('Data type:', type(new_df))
print('Updated data shape:', new_df.shape)
print(new_df['math level'].value_counts())
```

## 4. Save the updated CSV in the repository folder


```python
output_path = repo_path / 'data' / 'Student Performance Updated.csv'
new_df.to_csv(output_path, index=False)
print('Saved locally in the cloned repository:', output_path)
print('Note: This does not upload the change to GitHub.')
```

5. Displayed counts of the two categories to help check the result.
