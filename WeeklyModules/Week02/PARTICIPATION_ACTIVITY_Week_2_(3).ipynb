```python
letters =  ['a',  'b',  'c', 'd']

print(letters [int(int('3'*2)//11)])
# '3' * 2 becomes '33', then 33 // 11 = 3, so letters[3] is 'd'.

print(letters[-1])
# -1 refers to the last item in the list, so it evaluates to 'd'.

print(letters[:2])
# [:2] returns the items before index 2, so it evaluates to ['a',  'b']
```

```python
spam = ['apples', 'bananas', 'tofu', 'cats']
def comma_code(items):
  if len(items)==0:
      return ''
  elif len(items)==1:
      return items[0]
  else:
      return ', '. join (items[:-1]) + ', and ' + items[-1]
comma_code(spam)
```
